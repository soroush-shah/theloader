# theloader on Google Colab
Downloads the high-resolution textures of your site into category folders, saved on **Google Drive**.

Run the cells top to bottom (▶). Only cell 1 needs editing.

```
#@title 1) Settings
SITE      = "https://example.com"   #@param {type:"string"}
IDS       = "1-50"                  #@param {type:"string"}
MIN_SIZE  = 600                     #@param {type:"integer"}
DRIVE_DIR = "/content/drive/MyDrive/textures"  #@param {type:"string"}
EXTRA     = ""                      #@param {type:"string"}
# IDS: "1-500" or "3,7,10-20". Leave empty to find pages automatically (sitemap / crawl).
# EXTRA: any additional options, e.g. --category-selector ".category a" --best-only
```

```
#@title 2) Install
!pip -q install requests beautifulsoup4 Pillow
import os, subprocess
REPO = "soroush-shah/theloader"
token = ""
try:
    from google.colab import userdata
    token = userdata.get("GITHUB_TOKEN")   # only needed if the repo is private
except Exception:
    pass
url = f"https://{token}@github.com/{REPO}.git" if token else f"https://github.com/{REPO}.git"
if not os.path.exists("/content/theloader"):
    subprocess.run(["git", "clone", "-q", "--depth", "1", url, "/content/theloader"], check=True)
print("ready")
```

```
#@title 3) Connect Google Drive (files are saved there)
from google.colab import drive
drive.mount("/content/drive")
```

```
#@title 4) Test: analyse the first 3 pages (nothing is downloaded)
ids = f"--ids {IDS}" if IDS.strip() else "--crawl"
!python /content/theloader/theloader.py "{SITE}" {ids} --limit 3 --probe --min-size {MIN_SIZE} {EXTRA}
```

```
#@title 5) Download everything
ids = f"--ids {IDS}" if IDS.strip() else ""
!python /content/theloader/theloader.py "{SITE}" {ids} --min-size {MIN_SIZE} --out "{DRIVE_DIR}" {EXTRA}
```

```
#@title 6) (optional) Make one ZIP file and download it to your computer
import shutil
from google.colab import files
zip_path = shutil.make_archive("/content/textures", "zip", DRIVE_DIR)
files.download(zip_path)
```
